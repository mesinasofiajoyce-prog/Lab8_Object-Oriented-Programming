```python
class Book:
    def __init__(self, title):
        self.title = title
        self.available = True

class Library:
    def __init__(self):
        self.books = []

    def add(self, title):
        self.books.append(Book(title))

    def borrow(self, title):
        for b in self.books:
            if b.title == title:
                if b.available:
                    b.available = False
                    print("Borrowed")
                else:
                    print("Already borrowed")
                return
        print("Not found")

    def return_book(self, title):
        for b in self.books:
            if b.title == title:
                b.available = True
                print("Returned")

lib = Library()

for _ in range(int(input("Books: "))):
    lib.add(input("Title: "))

lib.borrow(input("Borrow: "))
lib.return_book(input("Return: "))
```

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock


class ShoppingCart:
    def __init__(self):
        self.cart = {}
        self.total = 0

    def add_product(self, product, quantity):
        if quantity <= 0:
            return "Invalid quantity."

        if quantity > product.stock:
            return f"Not enough stock for {product.name}. Available: {product.stock}"

        product.stock -= quantity

        if product.name in self.cart:
            self.cart[product.name]["quantity"] += quantity
        else:
            self.cart[product.name] = {"price": product.price, "quantity": quantity}

        self.total += product.price * quantity
        return f"{quantity} {product.name}(s) added to cart."

    def remove_product(self, product, quantity):
        if product.name not in self.cart:
            return "Product not in cart."

        if quantity > self.cart[product.name]["quantity"]:
            return "Cannot remove more than in cart."

        self.cart[product.name]["quantity"] -= quantity
        product.stock += quantity
        self.total -= product.price * quantity

        if self.cart[product.name]["quantity"] == 0:
            del self.cart[product.name]

        return f"{quantity} {product.name}(s) removed from cart."

    def view_cart(self):
        if not self.cart:
            return "Cart is empty."

        output = "Cart Contents:\n"
        for name, info in self.cart.items():
            output += f"- {name}: {info['quantity']} x {info['price']} = {info['quantity'] * info['price']}\n"
        output += f"Total Price: {self.total}"
        return output


# ---------------- MAKEUP PRODUCTS ----------------
products = {
    "Lipstick": Product("Lipstick", 299, 10),
    "Foundation": Product("Foundation", 499, 8),
    "Mascara": Product("Mascara", 350, 12),
    "Blush": Product("Blush", 275, 15),
    "Eyeliner": Product("Eyeliner", 199, 20)
}

cart = ShoppingCart()

# ---------------- MENU SYSTEM ----------------
while True:
    print("\n--- MAKEUP SHOPPING CART ---")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. Remove from Cart")
    print("4. View Cart")
    print("5. Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        print("\nAvailable Products:")
        for p in products.values():
            print(f"{p.name} - ₱{p.price} (Stock: {p.stock})")

    elif choice == "2":
        name = input("Enter product name: ")
        qty = int(input("Enter quantity: "))

        if name in products:
            print(cart.add_product(products[name], qty))
        else:
            print("Invalid product selection.")

    elif choice == "3":
        name = input("Enter product name to remove: ")
        qty = int(input("Enter quantity: "))

        if name in products:
            print(cart.remove_product(products[name], qty))
        else:
            print("Invalid product selection.")

    elif choice == "4":
        print(cart.view_cart())

    elif choice == "5":
        print("Thank you for shopping!")
        break

    else:
        print("Invalid choice.")
```
